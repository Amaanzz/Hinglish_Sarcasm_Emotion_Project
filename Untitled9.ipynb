```python
import pandas as pd

df = pd.read_csv("hinglish_augmented_175.csv")
df = df[['text','sarcasm']]
df['text'] = df['text'].astype(str)
df['sarcasm'] = df['sarcasm'].astype(int)

df.head(), df.shape
```

```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(df, test_size=0.1, random_state=42)

len(train_df), len(test_df)
```

```python
from huggingface_hub import login
login()
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification

model_name = "ai4bharat/indic-bert"

tokenizer = AutoTokenizer.from_pretrained(model_name, use_auth_token=True)
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2, use_auth_token=True)
```

```python
from datasets import Dataset
from transformers import DataCollatorWithPadding

train_ds = Dataset.from_pandas(train_df)
test_ds = Dataset.from_pandas(test_df)

def tokenize(batch):
    return tokenizer(batch["text"], padding="max_length", truncation=True, max_length=64)

train_ds = train_ds.map(tokenize, batched=True, batch_size=32, remove_columns=["text"])
test_ds = test_ds.map(tokenize, batched=True, batch_size=32, remove_columns=["text"])

train_ds = train_ds.rename_column("sarcasm", "labels")
test_ds = test_ds.rename_column("sarcasm", "labels")

data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

train_ds, test_ds, data_collator
```

```python
from transformers import TrainingArguments
import torch

batch = 32 if torch.cuda.is_available() else 8

training_args = TrainingArguments(
    output_dir="./sarcasm_finetuned",
    eval_strategy="epoch",
    save_strategy="epoch",
    learning_rate=2e-5,
    num_train_epochs=2,
    per_device_train_batch_size=batch,
    per_device_eval_batch_size=batch,
    weight_decay=0.01,
    load_best_model_at_end=True,
    logging_steps=100,
    push_to_hub=True,
    hub_model_id="amaan00z/hinglish-sarcasm-ft",
)
```

```python
from transformers import Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_ds,
    eval_dataset=test_ds,
    data_collator=data_collator,
)
```

```python
trainer.train()
```

```python
trainer.push_to_hub()
tokenizer.push_to_hub("amaan00z/hinglish-sarcasm-ft")
```
